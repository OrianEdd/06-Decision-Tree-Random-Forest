## Klasifikasi dengan Decision Tree dan Random Forest

Untuk melakukan klasifikasi, kita akan menggunakan dataset `03_EWC2025_Club_Championship_Standings.csv`.

```
import pandas as pd

# Load the dataset
file_path = '/content/03_EWC2025_Club_Championship_Standings.csv'
df = pd.read_csv(file_path)

# Display the first 5 rows and information about the dataframe
print("Dataset Head:")
import IPython.display
IPython.display.display(df.head())

print("\nDataset Info:")
df.info()
```

Dari kolom yang tersedia, `Position` akan digunakan sebagai target variabel. Kolom kategorikal seperti `Team_Name`, `Region`, dan `Status` akan diubah menjadi numerik menggunakan One-Hot Encoding. Kolom `Points_Total`, `Points_Tournament`, dan `Points_Bonus` akan digunakan sebagai fitur.

```
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report
import pandas as pd

# Drop columns that are too complex for a basic classification model
df_processed = df.drop(columns=['Games_Won'], errors='ignore')

# One-Hot Encode categorical columns: Organization, Region, Club_Partner_Status, Eligible_to_Win
categorical_cols = ['Organization', 'Region', 'Club_Partner_Status', 'Eligible_to_Win']
df_processed = pd.get_dummies(df_processed, columns=categorical_cols, drop_first=True)

# Target: 'Rank' (predicting the specific rank level)
# Features: All other columns
X = df_processed.drop('Rank', axis=1)
y = df_processed['Rank']

# Fill any missing values with 0
X = X.fillna(0)

# Split the dataset into 70% Training and 30% Testing
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

print('Preprocessing completed successfully.')
print(f'Total features after encoding: {X.shape[1]}')
print(f'Training set size: {len(X_train)}')
print(f'Test set size: {len(X_test)}')
```

### Klasifikasi menggunakan Decision Tree

```
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import classification_report, accuracy_score

# Membangun dan melatih model Decision Tree
dt_model = DecisionTreeClassifier(random_state=42)
dt_model.fit(X_train, y_train)

# Prediksi menggunakan data uji
y_pred_dt = dt_model.predict(X_test)

# Evaluasi model
print("--- Decision Tree Classification Report ---")
print(classification_report(y_test, y_pred_dt))
print(f"Akurasi Decision Tree: {accuracy_score(y_test, y_pred_dt):.2f}")
```

### Klasifikasi menggunakan Random Forest

```
from sklearn.ensemble import RandomForestClassifier

# Membangun dan melatih model Random Forest
rf_model = RandomForestClassifier(n_estimators=100, random_state=42)
rf_model.fit(X_train, y_train)

# Prediksi menggunakan data uji
y_pred_rf = rf_model.predict(X_test)

# Evaluasi model
print("--- Random Forest Classification Report ---")
print(classification_report(y_test, y_pred_rf))
print(f"Akurasi Random Forest: {accuracy_score(y_test, y_pred_rf):.2f}")
```

### Evaluasi Mendalam: Confusion Matrix

```
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report

def evaluate_model(y_true, y_pred, title):
    # Calculate metrics
    acc = accuracy_score(y_true, y_pred)
    report = classification_report(y_true, y_pred, zero_division=0)
    cm = confusion_matrix(y_true, y_pred)

    print(f"--- {title} ---")
    print(f"Accuracy: {acc:.2f}")
    print("Classification Report:")
    print(report)

    # Plot Confusion Matrix
    plt.figure(figsize=(8, 6))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                xticklabels=sorted(y_true.unique()),
                yticklabels=sorted(y_true.unique()))
    plt.title(f'Confusion Matrix - {title}')
    plt.xlabel('Predicted Label')
    plt.ylabel('True Label')
    plt.show()

# Evaluate Decision Tree
evaluate_model(y_test, y_pred_dt, "Decision Tree")

# Evaluate Random Forest
evaluate_model(y_test, y_pred_rf, "Random Forest")
```

**Mengapa keduanya mendapatkan akurasi 0?**
1. **Target Terlalu Unik:** Variabel `Rank` memiliki nilai yang berbeda untuk setiap baris (1, 2, 3, ...). Model menganggap setiap peringkat sebagai kategori yang terpisah sepenuhnya.
2. **Ukuran Data:** Dengan hanya 24 baris data, model tidak memiliki cukup contoh untuk mempelajari hubungan antara fitur (seperti `Total_Points`) dengan `Rank` tertentu yang belum pernah dilihat di data uji.
3. **Data Uji yang Berbeda:** Data yang ada di *test set* memiliki nilai `Rank` yang tidak ada di *training set* untuk kombinasi fitur tertentu, sehingga model menebak kelas yang salah.

  Jawaban: Decision Tree rentan overfitting karena algoritma ini berusaha membuat aturan spesifik untuk setiap detail data latih, termasuk pola acak atau noise. Karena dibangun dari ratusan pohon yang dilatih secara acak, kesalahan spesifik dari satu pohon akan dikompensasi oleh prediksi dari pohon-pohon lainnya, sehingga menghasilkan varians yang rendah dan keumuman model (generalization) yang lebih baik.

3. Apa fungsi bagging dan pemilihan subset fitur acak pada Random Forest?

  Jawaban: Bagging (Bootstrap Aggregating): Berfungsi untuk menciptakan sampel data latih yang bervariasi (diambil secara acak dengan pengembalian) untuk setiap pohon. Ini memastikan setiap pohon belajar dari sudut pandang data yang sedikit berbeda. Pemilihan subset fitur acak: Berfungsi untuk memastikan bahwa setiap pohon di dalam hutan benar-benar unik. Jika tidak diacak, sebuah fitur yang sangat dominan akan selalu terpilih menjadi akar utama di semua pohon, membuat hasil antar pohon menjadi terlalu mirip (correlated).
